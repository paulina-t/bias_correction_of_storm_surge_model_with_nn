```python
feature_vars = [
    'obs', 
    'tide_py', 
    'stormsurge_corrected', 
    'u10', 
    'v10'
    ] 

label_var = 'stormsurge_corrected - (obs - tide_py)'

window_width_past = [24, 24, 0, 0, 0]
window_width_future = [-1, 60, 60, 60, 60]  # -1 does not include time t0
horizon = 60  

datetime_start_hourly = datetime_predict
```
